<h3>Multi-output MLP

https://machinelearningmastery.com/deep-learning-models-for-multi-output-regression/ 

```python
# mlp for multi-output regression
from numpy import mean
from numpy import std
from sklearn.datasets import make_regression
from sklearn.model_selection import RepeatedKFold
from keras.models import Sequential
from keras.layers import Dense
```

```python
def get_model(n_inputs, n_outputs):
	model = tf.keras.Sequential()
	model.add(layers.Dense(20, input_dim=n_inputs, kernel_initializer='he_uniform', activation='relu'))
	model.add(layers.Dense(n_outputs))
	model.compile(loss='mae', optimizer='adam')
	return model
```

```python
# evaluate a model using repeated k-fold cross-validation
def evaluate_model(X, y):
	results = list()
	n_inputs, n_outputs = X.shape[1], y.shape[1]
	# define evaluation procedure
	# cv = RepeatedKFold(n_splits=10, n_repeats=3, random_state=1)
	cv = RepeatedKFold(n_splits=2, n_repeats=3, random_state=1)
	# enumerate folds
	for train_ix, test_ix in cv.split(X):
		# prepare data
		X_train, X_test = X[train_ix], X[test_ix]
		y_train, y_test = y[train_ix], y[test_ix]
		# define model
		model = get_model(n_inputs, n_outputs)
		# fit model
		model.fit(X_train, y_train, verbose=0, epochs=100)
		# evaluate model on test set
		mae = model.evaluate(X_test, y_test, verbose=0)
		# store result
		print('>%.3f' % mae)
		results.append(mae)
	return results
```

<h4>Increase no. of neurons and no. of layers</h4>

```python
def get_model_1(n_inputs, n_outputs):
	model = tf.keras.Sequential()
	model.add(layers.Dense(64, input_dim=n_inputs, kernel_initializer='he_uniform', activation='relu'))
	model.add(layers.Dense(64, input_dim=64, kernel_initializer='he_uniform', activation='relu'))
	model.add(layers.Dense(n_outputs))
	model.compile(loss='mae', optimizer='adam')
	return model
```

Tried with 100 epochs (not saved) - not much better.
Increased to 500 epochs - now the predicted values are at least sensible values.

```python
data_concat_xy = data_concat[["s0", "s1", "s2", "x", "y"]].dropna()
train_dataset = data_concat_xy.sample(frac=0.8, random_state=0)
test_dataset = data_concat_xy.drop(train_dataset.index)

x = train_dataset[["s0", "s1", "s2"]].to_numpy()
y = train_dataset[["x", "y"]].to_numpy()
#one training set and predict using test data
model = get_model_1(x.shape[1], y.shape[1])
# fit the model on all data
model.fit(x, y, verbose=0, epochs=500)
y_pred = model.predict(test_dataset[["s0", "s1", "s2"]].to_numpy())
fig, axs = plt.subplots(1,2)
axs[0].scatter(test_dataset["x"], test_dataset["y"])
axs[1].scatter(y_pred[:, 0], y_pred[:, 1])
```

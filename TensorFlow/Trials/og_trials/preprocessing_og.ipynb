```python
import pandas as pd
import matplotlib.pyplot as plt
# import json
import numpy as np

def read_the_data(scan: str): 
    data = pd.read_csv(f"../../DeltabotController/python/scans/scan_{scan}/data.csv")
    # cx = data["cx"]
    # cy = data["cy"]

    # with open(f"../../DeltabotController/python/scans/scan_{scan}/meta.json") as f:
    #     jsn = json.load(f)

    # hough_params = jsn["viewer"]["hough_params"]

    # print(hough_params)
    # print(jsn)
    # display(data.head(10))
    return (data)
```

```python
def relative_cx_cy(data): #if this is done to every scan, there is no concept of the history and lack of homing - maybe get the centre of the first scan and remove from all??
    centre = data.loc[data["x"]==0] #or do we shift by the mean? 
    data["cx_rel"] = data["cx"] - centre["cx"].values[0]
    data["cy_rel"] = data["cy"] - centre["cy"].values[0]

def px_to_mm(data, x_conversion, y_conversion): 
    data["cx_mm"] = data["cx_rel"]*x_conversion
    data["cy_mm"] = data["cy_rel"]*y_conversion

def combine_the_data(scans: list[(str, bool)], x_conversion, y_conversion):
    df_list = []
    for scan in scans:
        data = read_the_data(scan[0])
        relative_cx_cy(data)   
        px_to_mm(data, x_conversion, y_conversion)
        if scan[1]:
            data["cx_mm"] = -data["cx_mm"]
        df_list.append(data)
        plt.plot(data["cx_mm"], data["s1"], 'o', label=f"{scan[0]}")
        plt.legend()
    combined_data = pd.concat(df_list,ignore_index=True)

    return combined_data

x_px = 214.91 - 196.08 #0.1mm
y_px = 45.12 - 25.04 #0.1mm
x_conversion = 0.1/x_px #px to mm conversion 
y_conversion = 0.1/y_px #px to mm conversion 
data = combine_the_data([("0014_dscan", True),("0017_dscan", True),("0019_dscan", False),("0020_dscan", False),("0024_dscan", False),("0044_dscan", True)], x_conversion=x_conversion, y_conversion=y_conversion)
display(data[["x", "cx_mm", "y", "cy_mm"]])

```

<h3>Linear Regression

```python
from tensorflow.keras import layers

cx = np.array(train_features['cx_mm'])

cx_normalizer = layers.Normalization(input_shape=[1,], axis=None)
cx_normalizer.adapt(cx)
```

```python
cx_model = tf.keras.Sequential([
    cx_normalizer,
    layers.Dense(units=1)
])

cx_model.summary()
```

```python
cx_model.predict(cx[:10])
```

```python
cx_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.1),
    loss='mean_absolute_error')
```

```python
history = cx_model.fit(
    train_features['cx_mm'],
    train_labels,
    epochs=100,
    # Suppress logging.
    verbose=0,
    # Calculate validation results on 20% of the training data.
    validation_split = 0.2)
```

```python
hist = pd.DataFrame(history.history)
hist['epoch'] = history.epoch
hist.tail()
```

```python
def plot_loss(history):
  plt.plot(history.history['loss'], label='loss')
  plt.plot(history.history['val_loss'], label='val_loss')
  plt.ylim([0, 1100])
  plt.xlabel('Epoch')
  plt.ylabel('Error [MPG]')
  plt.legend()
  plt.grid(True)
plot_loss(history)
```

```python
test_results = {}

test_results['cx_model'] = cx_model.evaluate(
    test_features['cx_mm'],
    test_labels, verbose=0)
```

```python
x = tf.linspace(-1.5, 1.5, 251)
y = cx_model.predict(x)
```

```python
def plot_cx(x, y):
  plt.scatter(train_features['cx_mm'], train_labels, label='Data')
  plt.plot(x, y, color='k', label='Predictions')
  plt.xlabel('cx_mm')
  plt.ylabel('s1')
  plt.legend()
plot_cx(x, y)
```

```python
test_results['linear_model'] = linear_model.evaluate(
    test_features, test_labels, verbose=0)
```

<h3>DNN

```python
def build_and_compile_model(norm):
  model = tf.keras.Sequential([
      norm,
      layers.Dense(64, activation='relu'),
      layers.Dense(64, activation='relu'),
      layers.Dense(1)
  ])

  model.compile(loss='mean_absolute_error',
                optimizer=tf.keras.optimizers.Adam(0.001))
  return model
```

```python
dnn_cx_model = build_and_compile_model(cx_normalizer)
dnn_cx_model.summary()
```

```python
history = dnn_cx_model.fit(
    train_features['cx_mm'],
    train_labels,
    validation_split=0.2,
    verbose=0, epochs=100)
```

```python
plot_loss(history)
```

```python
x = tf.linspace(-2, 2, 251)
y = dnn_cx_model.predict(x)
plot_cx(x, y)
```

```python
test_results['dnn_cx_model'] = dnn_cx_model.evaluate(
    test_features['cx_mm'], test_labels,
    verbose=0)

```

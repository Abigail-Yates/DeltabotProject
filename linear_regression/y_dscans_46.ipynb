```python
scan = "0048_dscan"
data, cx, cy, hough_params =  lr.read_the_data(scan)
```

```python
lr.subplot_linear(scan, hough_params, cx, cy, data["radius"], 100, 500, True)
lr.single_plot(scan, hough_params, cy, cx,  data["radius"], None, None, True)
```
